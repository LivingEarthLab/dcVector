<h1> Vectors data within the cube in a box </h1>
<p> You can use this notebook to test the setup of your cube in a box to handle vectors data </p>
<p> This notebook is localised in the Narbonne region, Occitanie, France and uses a subset of the land-cover data (ocs) of the Grand Narbonne and the Parc naturel régional de la Narbonnaise en Méditerranée, produced by Le Grand Narbonne Communauté d'Agglomération (ODbL v1.0) </p>
<p> Please ensure that those data are indexed within the datacube (see readme) </p>

```python
import datacube
from datacube.vectors import Legend, TemporalMapViewer, create_shape_geometry, Load, create_linear_cmap
```

<p> The query represents the ROI and the period of interest </p>
<p> <i> A specific list of Ids and field name could also be specified for vectors</i></p>

```python
geom_inter={"geometry": { "type": "Polygon", "coordinates": [ [ [ 2.901570798515546, 43.291865029294144 ], [ 2.885464555159261, 43.27445129345984 ], [ 2.92260708095458, 43.237532582716518 ], [ 2.961922015424769, 43.261564880808827 ], [ 2.935327810238215, 43.292681313355232 ], [ 2.901570798515546, 43.291865029294144 ] ] ] } }
ids=[5624,14107,14108,17032,17036,19103,29204,36561,37143,37249,37252]
query= {
    'time': ['2000-01-01', '2024-12-31'],
    'crs':'EPSG:2154',
    'output_crs': 'epsg:2154',
    "product":"ocs",
    #'spatialFilter':geom_inter,
    #"spatialFilter_crs":"epsg:4326",
    #"ids":ids
}
```

<h3> Download of the data using the query </h3>
<p> All data with a valid product name in the query will be downloaded and stored into a unique instance of the Load class. Information and data can be retrieved and displayed from this object. </p>

```python
results=Load(query)
```

```python
#Retrieving vectors data as a Panda dataframe
data_vector=results.get_vector_dataset
data_vector
```

```python
#Plotting the vector time series
ax_vect,fig_vect=results.vector_subplot("niveau 4")
```

<h3> Vectors data could be displayed into dynamic maps </h3>

```python
default_legend=Legend(data_vector.attrs['legend'])
colorscheme=default_legend.get_colorschemeByName('niveau 4')
field=default_legend.get_colorscheme_field('niveau 4')[0]
legend=default_legend.get_legendFromColorscheme('niveau 4')
viewer = TemporalMapViewer(gdf=data_vector, legendDict=colorscheme,date_column='time',id_column='id',display_column=field,legend=legend)
```
